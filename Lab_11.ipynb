# Travelling Salesman Problem using Dynamic Programming.
Problem Description:
Given a set of cities and distances, find the shortest possible route that visits each city once and returns to the starting city.

#Theory
TSP uses dynamic programming with bit masking. Store results of visited subsets to reduce repeated work.

## Algorithm
1. Start from source city
2. Mark visited cities using bit mask
3. Try all unvisited cities
4. Choose minimum cost path
5. Return to source city

```python
def tsp(graph):
    n = len(graph)
    dp = [[-1]*(1<<n) for _ in range(n)]
    parent = [[-1]*(1<<n) for _ in range(n)] # To store the next city in the optimal path

    def visit(city, mask):
        if mask == (1<<n) - 1:
            # If all cities visited, return cost to return to start city (city 0)
            return graph[city][0]

        if dp[city][mask] != -1:
            return dp[city][mask]

        ans = float('inf')
        best_next_city = -1

        for next_city_cand in range(n):
            # Check if next_city_cand is not visited yet
            if mask & (1<<next_city_cand) == 0: # 0 means not visited
                current_cost = graph[city][next_city_cand] + visit(next_city_cand, mask | (1<<next_city_cand))
                if current_cost < ans:
                    ans = current_cost
                    best_next_city = next_city_cand

        dp[city][mask] = ans
        parent[city][mask] = best_next_city # Store the chosen next city for path reconstruction
        return ans

    min_cost = visit(0, 1) # Start from city 0, mask 1 (only city 0 visited)

    # Reconstruct the path
    path = []
    current_city = 0
    current_mask = 1
    path.append(current_city)

    while current_mask != (1<<n) - 1:
        next_city = parent[current_city][current_mask]
        if next_city == -1: # This should not happen if a path exists
            break
        path.append(next_city)
        current_mask |= (1<<next_city) # Add next_city to visited mask
        current_city = next_city

    path.append(0) # Return to the starting city

    return min_cost, path
```

```python
def get_user_graph():
    while True:
        try:
            n = int(input("Enter the number of cities: "))
            if n <= 1:
                print("Number of cities must be at least 2.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter an integer.")

    graph = []
    print(f"Enter the {n}x{n} distance matrix (row by row). Enter 'inf' for no direct path. Separate values by spaces:")
    for i in range(n):
        while True:
            try:
                row_input = input(f"Enter row {i}: ").split()
                if len(row_input) != n:
                    print(f"Expected {n} values, but got {len(row_input)}. Please re-enter the row.")
                    continue
                row = []
                for val in row_input:
                    if val.lower() == 'inf':
                        row.append(float('inf'))
                    else:
                        row.append(int(val))
                graph.append(row)
                break
            except ValueError:
                print("Invalid input. Please enter integers or 'inf'.")
    return graph
```

```python
# Take graph input from the user
user_graph = get_user_graph()

# Calculate TSP with the user's graph
min_cost, path = tsp(user_graph)

print("Minimum travelling cost:", min_cost)
print("Path:", ' -> '.join(map(str, path)))
```
