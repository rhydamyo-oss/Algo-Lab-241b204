# Shortest Path Using Dijkstra Algorithm in Python
Problem Description:
Given a weighted graph with non negative edge weights. Find the shortest distance from a source vertex to all other vertices using Dijkstra algorithm

#Theory
Dijkstra algorithm finds shortest distance from one source node to all other nodes in a weighted graph with non negative weights. Dictionary structure stores nodes as keys and neighbor nodes with weights as values.

##Algorithm Steps

1. Represent graph using dictionary.
2. Set distance of every node as infinity.
3. Set distance of source node as 0.
4. Select node with smallest distance among unvisited nodes.
5. Update distance of neighbor nodes.
6. Repeat until all nodes get visited.

```python
def dijkstra(graph, start):

    distances = {}
    predecessors = {}
    visited = []

    for node in graph:
        distances[node] = float('inf')
        predecessors[node] = None

    distances[start] = 0

    while len(visited) < len(graph):

        min_node = None
        min_distance = float('inf')

        for node in graph:
            if node not in visited and distances[node] < min_distance:
                min_distance = distances[node]
                min_node = node

        if min_node is None: # All remaining nodes are unreachable
            break

        visited.append(min_node)

        for neighbor, weight in graph[min_node].items():
            new_distance = distances[min_node] + weight

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                predecessors[neighbor] = min_node

    return distances, predecessors
```

```python
graph = {
'A': {'B': 4, 'C': 2},
'B': {'A': 4, 'C': 5, 'D': 10},
'C': {'A': 2, 'B': 5, 'D': 3},
'D': {'B': 10, 'C': 3}
}

result = dijkstra(graph, 'A')

for node in result:
    print("Distance from A to", node, "=", result[node])
```

## Extended Graph and Path Reconstruction

```python
# Add two more nodes to the graph
extended_graph = {
    'A': {'B': 4, 'C': 2},
    'B': {'A': 4, 'C': 5, 'D': 10, 'E': 7},
    'C': {'A': 2, 'B': 5, 'D': 3},
    'D': {'B': 10, 'C': 3, 'F': 6},
    'E': {'B': 7, 'F': 8},
    'F': {'D': 6, 'E': 8}
}

def get_shortest_path(predecessors, start_node, target_node):
    path = []
    current = target_node
    while current is not None:
        path.insert(0, current)
        current = predecessors.get(current)

    # Check if a path exists from start_node to target_node
    if path[0] == start_node:
        return ' -> '.join(path)
    else:
        return f"No path from {start_node} to {target_node}"
```

## Test with User Input and Display Paths

```python
print("Available nodes in the graph:", list(extended_graph.keys()))
source_node = input("Enter the source node: ").upper()

if source_node not in extended_graph:
    print(f"Error: Node '{source_node}' not found in the graph.")
else:
    distances, predecessors = dijkstra(extended_graph, source_node)

    print(f"\nShortest distances from {source_node}:")
    for node, dist in distances.items():
        if dist == float('inf'):
            print(f"  To {node}: Unreachable")
        else:
            print(f"  To {node}: {dist}")

    print(f"\nShortest paths from {source_node}:")
    for node in extended_graph:
        if node != source_node:
            path = get_shortest_path(predecessors, source_node, node)
            print(f"  Path to {node}: {path}")
        else:
            print(f"  Path to {node}: {source_node}")
```
