# Greedy Knapsack Problem (Fractional)
In this lab, we explore the Fractional Knapsack Problem, a fundamental optimization problem that demonstrates the Greedy Algorithmic Paradigm.
We will implement it using greedy selection (based on value-to-weight ratio) and analyze its time and space complexity.

- Created by Dr. Ajay

# Problem Definition
Given a set of
n
n items, each with:
Value vi
Weight wi
A knapsack with maximum capacity **W**

Goal: maximize total value that can be put in the knapsack.
Constraint: You may take fractions of items.

Greedy Strategy:

*  Sort items by decreasing ratio vi/wi
*  Select items fully until knapsack is full.
*  Take fraction of the next item if capacity remains.

```python
def fractional_knapsack(capacity, items):


    # Calculate value-to-weight ratio for each item and store with original index
    for i, item in enumerate(items):
        item['ratio'] = item['value'] / item['weight']
        item['original_index'] = i

    # Sort items by decreasing value-to-weight ratio
    items.sort(key=lambda x: x['ratio'], reverse=True)

    total_value = 0.0
    knapsack_contents = []

    for item in items:
        if capacity == 0:
            break

        if item['weight'] <= capacity:
            # Take the whole item
            total_value += item['value']
            capacity -= item['weight']
            knapsack_contents.append({'item_index': item['original_index'], 'fraction': 1.0, 'value': item['value'], 'weight': item['weight']})
        else:
            # Take a fraction of the item
            fraction = capacity / item['weight']
            total_value += item['value'] * fraction
            knapsack_contents.append({'item_index': item['original_index'], 'fraction': fraction, 'value': item['value'] * fraction, 'weight': item['weight'] * fraction})
            capacity = 0  # Knapsack is full
            break

    return total_value, knapsack_contents

# Example Usage:
# Define items as dictionaries with 'value' and 'weight'
items = [
    {'value': 60, 'weight': 10},
    {'value': 100, 'weight': 20},
    {'value': 120, 'weight': 30}
]
knapsack_capacity = 50

max_value, selected_items = fractional_knapsack(knapsack_capacity, items)

print(f"Knapsack Capacity: {knapsack_capacity}")
print(f"Maximum Total Value: {max_value:.2f}")
print("Selected Items:")
for item in selected_items:
    print(f"  Item {item['item_index']}: Fraction = {item['fraction']:.2f}, Value = {item['value']:.2f}, Weight = {item['weight']:.2f}")
```

In this lab, we implement Huffman Coding, a greedy algorithm used for lossless data compression.
It assigns variable-length binary codes to input symbols based on their frequencies, ensuring that more frequent symbols receive shorter codes and less frequent symbols receive longer codes.

This algorithm constructs a binary tree (Huffman Tree) where each leaf node represents a character, and the path from the root to a leaf gives the character’s binary code.

Use:
Huffman Tree Construction using a Min-Priority Queue (Min-Heap).
Code generation for each character.
Encoding and decoding using the generated Huffman codes.
Complexity analysis and comparison with fixed-length encoding.

# Pseudocode:
1. Create a leaf node for each character and insert all nodes into a
2. min-heap by frequency.
   *  While the heap has more than one node:
   * Extract two nodes with the lowest frequencies.
   * Create a new internal node with frequency = sum of the two.
   *  Set the two extracted nodes as left and right children.
   *  Insert the new node back into the min-heap.
3. The remaining node is the root of the Huffman Tree.

4. Traverse the tree:
   * Assign 0 for left edge and 1 for right edge.
   * The resulting paths from root to leaves give Huffman codes.

```python
import heapq
from collections import defaultdict

# 1. Huffman Node Class
class HuffmanNode:
    def __init__(self, char, freq):
        self.char = char
        self.freq = freq
        self.left = None
        self.right = None

    # Used for min-heap comparison
    def __lt__(self, other):
        return self.freq < other.freq

# 2. Build Huffman Tree Function
def build_huffman_tree(text):
    # Calculate frequency of each character
    frequency = defaultdict(int)
    for char in text:
        frequency[char] += 1

    # Create a min-heap of HuffmanNodes
    min_heap = []
    for char, freq in frequency.items():
        heapq.heappush(min_heap, HuffmanNode(char, freq))

    # Build the tree
    while len(min_heap) > 1:
        node1 = heapq.heappop(min_heap)
        node2 = heapq.heappop(min_heap)

        # Create new internal node
        merged = HuffmanNode(None, node1.freq + node2.freq)
        merged.left = node1
        merged.right = node2
        heapq.heappush(min_heap, merged)

    return min_heap[0] if min_heap else None

# 3. Generate Huffman Codes Function
def generate_huffman_codes(node, current_code="", codes={}):
    if node is None:
        return

    # If it's a leaf node, assign the code
    if node.char is not None:
        codes[node.char] = current_code if current_code else "0" # Handle single character case
        return

    # Traverse left (0) and right (1)
    generate_huffman_codes(node.left, current_code + "0", codes)
    generate_huffman_codes(node.right, current_code + "1", codes)
    return codes

# 4. Encoding Function
def encode_text(text, codes):
    encoded_text = ""
    for char in text:
        encoded_text += codes[char]
    return encoded_text

# 5. Decoding Function
def decode_text(encoded_text, root):
    decoded_text = ""
    current_node = root
    for bit in encoded_text:
        if bit == '0':
            current_node = current_node.left
        else:
            current_node = current_node.right

        if current_node.char is not None:
            decoded_text += current_node.char
            current_node = root # Reset to root for next character
    return decoded_text

# Example Usage:
input_text = "this is an example for huffman encoding"
print(f"Original text: {input_text}")

# Build Huffman Tree
huffman_tree_root = build_huffman_tree(input_text)

# Generate Huffman Codes
huffman_codes = generate_huffman_codes(huffman_tree_root)
print(f"Huffman Codes: {huffman_codes}")

# Encode the text
encoded_data = encode_text(input_text, huffman_codes)
print(f"Encoded text: {encoded_data}")

# Decode the text
decoded_data = decode_text(encoded_data, huffman_tree_root)
print(f"Decoded text: {decoded_data}")

# Verification
if input_text == decoded_data:
    print("Encoding and decoding successful!")
else:
    print("Error: Decoded text does not match original.")

# Compare with fixed-length encoding
fixed_length_bits = len(input_text) * 8 # Assuming 8 bits per character (ASCII)
huffman_bits = len(encoded_data)

print(f"\nFixed-length encoding bits: {fixed_length_bits}")
print(f"Huffman encoding bits: {huffman_bits}")
print(f"Space saved: {fixed_length_bits - huffman_bits} bits")
```
